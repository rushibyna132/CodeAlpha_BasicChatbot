```python
def chatbot():
    print("🤖 Hello! I'm your chatbot.")
    print("You can chat with me, or type 'bye' to exit.\n")

    while True:
        user_input = input("You: ").lower().strip()

        if any(word in user_input for word in ["hello", "hi", "hii", "hey", "heyy"]):
            print("Bot: Hey! 😊 Nice to meet you!")

        elif "how are you" in user_input:
            print("Bot: I'm doing great! Thanks for asking. How are you?")

        elif "your name" in user_input or "who are you" in user_input:
            print("Bot: I'm a simple Python chatbot. You can call me ChatBot!")

        elif "thank" in user_input:
            print("Bot: You're very welcome! 😊")

        elif "good morning" in user_input:
            print("Bot: Good morning! 🌞 Have a wonderful day!")

        elif "good afternoon" in user_input:
            print("Bot: Good afternoon! 😊 How is your day going?")

        elif "good evening" in user_input:
            print("Bot: Good evening! 🌆 Hope you're having a nice day!")

        elif "bye" in user_input or "goodbye" in user_input:
            print("Bot: Goodbye! 👋 It was nice chatting with you!")
            break

        else:
            print("Bot: Hmm, I'm still learning! Try saying hello or asking how I am.")


chatbot()
```
